## Create a Function for reading in the 6 csv files
#### The function will also:
#### Rename columns
#### Create duplicate dataframe with required columns only
#### Create new columns for half time and full time goals
#### Create a Dictionary to replace League names from the csv files.  
#### The purpose of this is to stop code duplication

```python
def prepare_league(file_name): # file_name is a parameter given to the function
    
    # read in csv
    df = pd.read_csv(file_name)
    
    # rename columns
    df = df.rename(columns={
    "Div": "League",
    "HomeTeam": "Home_Team",
    "AwayTeam": "Away_Team",
    "FTHG": "FT_Home_Goals",
    "FTAG": "FT_Away_Goals",
    "FTR": "Result",
    "HTHG": "HT_Home_Goals",
    "HTAG": "HT_Away_Goals",
    "HST": "Home_Shots_On_Target",
    "AST": "Away_Shots_On_Target",
    } 
    )
    
    # create duplicate dataframe with required columns only
    df = df[
        ["Date", 
         "League", 
          "Home_Team", 
          "Away_Team", 
          "HT_Home_Goals",
          "HT_Away_Goals", 
          "FT_Home_Goals", 
          "FT_Away_Goals", 
          "Result"] 
    ].copy()
    
    # create new columns for half time and full time goals
    df["HT_Goals"] = (
    df["HT_Home_Goals"] + df["HT_Away_Goals"] 
    )

    df["FT_Goals"] = (
    df["FT_Home_Goals"] + df["FT_Away_Goals"] 
    )

    # Create a Dictionary to replace League names from the csv files.  This will make it more readable once i created a merged dataframe
    df["League"] = df["League"].replace( 
        {
            "D1": "German Bundesliga",
            "E0": "English Premier League",
            "E1": "English Championship",
            "SP1": "Spanish La Liga",
            "I1": "Italian Serie A",
            "F1": "French Ligue 1",
        }
    )
            

    return df
    
    
```
